## 3. Control flow

**Opdracht:** maak een function `geef_label(temp_max)` die `"hitte"`
teruggeeft bij 30 graden of meer, `"aangenaam"` tussen de 20 en 30 graden,
en anders `"fris"`.

````python
def geef_label(temp_max):
    if temp_max >= 30:
        return "hitte"
    elif temp_max >= 20:
        return "aangenaam"
    else:
        return "fris"

for dag in dagen[:5]:
    print(dag["datum"], geef_label(dag["temp_max_c"]))
````

## 5. Functions en scope

**Vraag:** wat gebeurt er als je `totaal` (gebruikt in de function
hieronder) na de aanroep buiten de function probeert te printen?

```python
def tel_op(lijst):
    totaal = 0
    for dag in lijst:
        totaal += dag["temp_max_c"]
    return totaal

tel_op(dagen)
print(totaal)
```

a) print het totaal
b) print 0
c) geeft een `NameError`
d) print `None`

Antwoord: **c**, `totaal` is een local variabele binnen `tel_op` en bestaat
niet buiten die scope.

````python
def tel_op(lijst):
    totaal = 0
    for dag in lijst:
        totaal += dag["temp_max_c"]
    return totaal

tel_op(dagen)

try:
    print(totaal)
except NameError as e:
    print("NameError:", e)
````

**Opdracht:** maak een function `analyseer_periode(periode)` die een dict
teruggeeft met de warmste en koudste datum uit die periode.

````python
def analyseer_periode(periode):
    warmste = periode[0]
    koudste = periode[0]

    for dag in periode:
        if dag["temp_max_c"] > warmste["temp_max_c"]:
            warmste = dag
        if dag["temp_max_c"] < koudste["temp_max_c"]:
            koudste = dag

    return {
        "warmste_datum": warmste["datum"],
        "koudste_datum": koudste["datum"],
    }

print(analyseer_periode(dagen))
````
